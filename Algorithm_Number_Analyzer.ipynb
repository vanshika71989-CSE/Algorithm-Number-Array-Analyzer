### 1. Prime Number Check

**Algorithm:**

1. Start
2. Enter a number n.
3. If n is less than 2, display "Not a Prime Number".
4. Otherwise, check whether n is divisible by any number from 2 to n-1.
5. If n is divisible by any number, display "Not a Prime Number".
6. Otherwise, display "Prime Number".
7. Stop

```python
n = int(input("Enter a number: "))

if n < 2:
    print("Not a Prime Number")
else:
    prime = True

    for i in range(2, n):
        if n % i == 0:
            prime = False
            break

    if prime:
        print("Prime Number")
    else:
        print("Not a Prime Number")
```

### 2. GCD (Greatest Common Divisor)

**Algorithm:**

1. Start
2. Enter two numbers a and b.
3. Find the remainder when a is divided by b.
4. Replace a with b and b with the remainder.
5. Repeat the process until b becomes 0.
6. The value of a is the GCD.
7. Display the GCD.
8. Stop

**Pseudocode:**

START

INPUT a, b

WHILE b ≠ 0
    
    remainder ← a MOD b
    a ← b
    b ← remainder
END WHILE

DISPLAY a

STOP

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

while b != 0:
    remainder = a % b
    a = b
    b = remainder

print("GCD =", a)
```

| 15, 10 | GCD = 5 | GCD = 5 |

### 3. Prime Factorization

**Algorithm:**

1. Start
2. Enter a number a.
3. Start checking from 2.
4. If a is divisible by the current number, print that number as a factor.
5. Divide a by that number.
6. Continue checking until a becomes 1.
7. Stop

**Pseudocode:**

START

INPUT a


SET i = 2

WHILE a > 1
   
    IF a MOD i = 0
        DISPLAY i
        a = a / i
    ELSE
        i = i + 1
    END IF

END WHILE

STOP

```python
a = int(input("Enter a number: "))

i = 2

while a > 1:
    if a % i == 0:
        print(i, end=" ")
        a = a // i
    else:
        i = i + 1
```

### 5. Palindrome

**Algorithm:**

1. Start
2. Enter a number n.
3. Store the original number in another variable.
4. Reverse the number using a loop.
5. Compare the reversed number with the original number.
6. If both are equal, display "Palindrome".
7. Otherwise, display "Not a Palindrome".
8. Stop

**Pseudocode:**

START

INPUT n

SET original = n

SET reverse = 0

WHILE n > 0
   
    digit = n MOD 10
    reverse = reverse * 10 + digit
    n = n // 10

END WHILE

IF reverse = original
   
    DISPLAY "Palindrome"

ELSE
   
    DISPLAY "Not a Palindrome"

END IF

STOP

```python
n = int(input("Enter a number: "))

original = n
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

if reverse == original:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

### 6. Fibonacci Series

**Algorithm:**

1. Start
2. Enter the number of terms n.
3. Set the first two terms as 0 and 1.
4. Display the terms one by one.
5. Find the third term by adding the previous two terms.
6. Repeat until n terms are displayed.
7. Stop

**Pseudocode:**

START

INPUT n

SET first = 0

SET second = 1

FOR i = 1 TO n
   
    DISPLAY first
    third = first + second
    first = second
    second = third

END FOR

STOP

```python
n = int(input("Enter the number of terms: "))

first = 0
second = 1

for i in range(n):
    print(first, end=" ")
    third = first + second
    first = second
    second = third
```

## Combined Number Analysis Module

```python
def prime_number(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True
```

```python
n = int(input("Enter a number: "))

if prime_number(n):
    print("Prime Number")
else:
    print("Not a Prime Number")
```

```python
def gcd(a, b):
    while b != 0:
        remainder = a % b
        a = b
        b = remainder

    return a
```

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("GCD =", gcd(a, b))
```

```python
def prime_factors(a):
    i = 2

    while a > 1:
        if a % i == 0:
            print(i, end=" ")
            a = a // i
        else:
            i = i + 1
```

```python
a = int(input("Enter a number: "))

print("Prime Factors:", end=" ")
prime_factors(a)
```

```python
def smallest_divisor(n):
    i = 2

    while n % i != 0:
        i = i + 1

    return i
```

```python
n = int(input("Enter a number: "))

print("Smallest Divisor =", smallest_divisor(n))
```

```python
def palindrome(n):
    original = n
    reverse = 0

    while n > 0:
        digit = n % 10
        reverse = reverse * 10 + digit
        n = n // 10

    return reverse == original
```

```python
n = int(input("Enter a number: "))

if palindrome(n):
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
def fibonacci(n):
    first = 0
    second = 1
    series = []

    for i in range(n):
        series.append(first)
        third = first + second
        first = second
        second = third

    return series
```

```python
n = int(input("Enter the number of terms: "))

print("Fibonacci Series:", fibonacci(n))
```

### 2.1 Maximum Element in an Array

#### Algorithm
1. Start
2. Input the number of elements in the array.
3. Input the array elements.
4. Assume the first element as maximum.
5. Compare each remaining element with maximum.
6. If an element is greater than maximum, update maximum.
7. Display the maximum element.
8. Stop

#### Pseudocode

START
Enter array elements

max ← first element

FOR each remaining element
   
    IF element > max THEN
        max ← element
    END IF

END FOR

Display max

STOP

```python
arr = list(map(int, input("Enter array elements: ").split()))

max_element = arr[0]

for i in range(1, len(arr)):
    if arr[i] > max_element:
        max_element = arr[i]

print("Maximum Element =", max_element)
```

### 2.2 Array Reversal

#### Algorithm

1. Start.
2. Enter the array elements.
3. Set the starting index to the first element and the ending index to the last element.
4. Swap the elements at the starting and ending positions.
5. Move the starting index forward and the ending index backward.
6. Repeat the process until the starting index is greater than or equal to the ending index.
7. Display the reversed array.
8. Stop.

#### Pseudocode

START

INPUT array elements

left ← 0

right ← length of array - 1

WHILE left < right
    
    SWAP array[left] and array[right]
    left ← left + 1
    right ← right - 1

END WHILE

OUTPUT array

STOP

```python
arr = list(map(int, input("Enter array elements: ").split()))

left = 0
right = len(arr) - 1

while left < right:
    arr[left], arr[right] = arr[right], arr[left]
    left = left + 1
    right = right - 1

print("Reversed Array =", arr)
```

### 2.4 Remove Duplicates

#### Algorithm

1. Start.
2. Enter the array elements.
3. Create an empty list to store unique elements.
4. Traverse each element of the array using a loop.
5. Check whether the element is already present in the unique list.
6. If the element is not present, add it to the unique list.
7. Display the array after removing duplicate elements.
8. Stop.

#### Pseudocode

START

INPUT array elements

unique ← empty list

FOR each element in the array
   
    IF element is not present in unique THEN
        ADD element to unique
    END IF

END FOR

OUTPUT unique

STOP

```python
arr = list(map(int, input("Enter array elements: ").split()))

unique = []

for i in range(len(arr)):
    if arr[i] not in unique:
        unique.append(arr[i])

print("Array after removing duplicates =", unique)
```

## Combined Array Analysis Module

```python
def maximum_element(arr):
    max_element = arr[0]

    for i in range(1, len(arr)):
        if arr[i] > max_element:
            max_element = arr[i]

    return max_element
```

```python
arr = list(map(int, input("Enter array elements: ").split()))

print("Maximum Element =", maximum_element(arr))
```

```python
def reverse_array(arr):
    left = 0
    right = len(arr) - 1

    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        left = left + 1
        right = right - 1

    return arr
```

```python
arr = list(map(int, input("Enter array elements: ").split()))

print("Reversed Array =", reverse_array(arr))
```

```python
def count_elements(arr):
    count = 0

    for i in range(len(arr)):
        count = count + 1

    return count
```

```python
arr = list(map(int, input("Enter array elements: ").split()))

print("Number of Elements =", count_elements(arr))
```

```python
def remove_duplicates(arr):
    unique = []

    for i in range(len(arr)):
        if arr[i] not in unique:
            unique.append(arr[i])

    return unique
```

```python
arr = list(map(int, input("Enter array elements: ").split()))

print("Array after removing duplicates =", remove_duplicates(arr))
```

```python
def kth_smallest(arr, k):
    arr.sort()

    return arr[k - 1]
```

```python
arr = list(map(int, input("Enter array elements: ").split()))
k = int(input("Enter the value of K: "))

print("Kth Smallest Element =", kth_smallest(arr, k))
```

### 3.4 Base Conversion

#### Algorithm
1. Start.
2. Enter a decimal number.
3. Enter the base to which the number has to be converted.
4. Repeatedly divide the number by the given base and store the remainders.
5. Continue until the number becomes 0.
6. Read the remainders in reverse order.
7. Display the converted number.
8. Stop.

#### Pseudocode

START

INPUT decimal number

INPUT base

result ← empty

number ← decimal number

WHILE number > 0
    
    remainder ← number MOD base
    ADD remainder to result
    number ← number DIV base

END WHILE

REVERSE result

OUTPUT result

STOP

```python
decimal = int(input("Enter a decimal number: "))
base = int(input("Enter the base: "))

number = decimal
result = ""

while number > 0:
    remainder = number % base

    if remainder < 10:
        result = str(remainder) + result
    else:
        result = chr(55 + remainder) + result

    number = number // base

print("Converted Number =", result)
```

## Combined Mathematical & Conversion Module

```python
def factorial_number(n):
    result = 1

    for i in range(1, n + 1):
        result = result * i

    return result
```

```python
n = int(input("Enter a number: "))

print("Factorial =", factorial_number(n))
```

```python
def power(base, exponent):
    result = 1

    for i in range(exponent):
        result = result * base

    return result
```

```python
base = int(input("Enter the base: "))
exponent = int(input("Enter the exponent: "))

print("Power =", power(base, exponent))
```

```python
import math

def square_root(n):
    return math.sqrt(n)
```

```python
n = float(input("Enter a number: "))

print("Square Root =", square_root(n))
```

```python
def base_conversion(decimal, base):
    number = decimal
    result = ""

    while number > 0:
        remainder = number % base

        if remainder < 10:
            result = str(remainder) + result
        else:
            result = chr(55 + remainder) + result

        number = number // base

    return result
```

```python
decimal = int(input("Enter a decimal number: "))
base = int(input("Enter the base: "))

print("Converted Number =", base_conversion(decimal, base))
```

```python
n = int(input("Enter a number: "))

factorial = 1

for i in range(1, n + 1):
    factorial = factorial * i

print("Factorial =", factorial)
```

```python
base = int(input("Enter the base: "))
exponent = int(input("Enter the exponent: "))

result = 1

for i in range(exponent):
    result = result * base

print("Power =", result)
```

```python
decimal = int(input("Enter a decimal number: "))
base = int(input("Enter the base: "))

number = decimal
result = ""

while number > 0:
    remainder = number % base

    if remainder < 10:
        result = str(remainder) + result
    else:
        result = chr(55 + remainder) + result

    number = number // base

print("Converted Number =", result)
```
